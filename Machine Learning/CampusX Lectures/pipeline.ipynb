### Steps
1. Raw data

2. Sperate columns on bases of numerical and categorical 
   - seperate categorical and numerical columns
   - apply preprocessing using Pipeline 
   - Syntax:
    ```
    numercal_pipeline = Pipeline([
        ("name", object), ("name", object)
    ])

    e.g.
        numerical_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy='mean')),
            ("scaler", StandardScaler(handle_unknown="ignore"))
        ])

    ```

    same for categorical columns

3. Combine both of numerical_pipeline and categorical_pipeline for transformation
   - Syntax
    ```
    preprocessing = ColumnTransformer([
        ("name", transformmer, columns),
        ("name", transformmer, columns)
    ])

    eg.

    preprocessing = ColumnTransformer([
        ("num", numerical_pipeline, numerical_cols),
        ("cat", categorical_pipeline, categorical_cols)
    ])
    ```
4. Make model 
    - Syntax:
        ```
        final_pipeline = PipeLine([
            ("name", Object),
            ("name", Object)
        ])

        e.g.

        final_pipeline = Pipeline([
            ("preprocessor", preprocessing),
            ("model", LogisticRegression())
        ])
        ```
5. Train model
    - Syntax
        ```
        final_model.fit(X_train, y_train)
        ``` 

6. Predict output
    - Syntax
        ```
        pred = final_pipeline(X_test)
        ```

````python
X = df.drop(["PassengerId", "Survived", "Name", "Cabin"], axis=1)
Y = df["Survived"]
````

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test  = train_test_split(X, Y, test_size=0.2, random_state=42)
````

````python
numerical_columns = ["Pclass", "Age", "SibSp", "Parch"]
categorical_columns = ["Sex", "Embarked"]
````

````python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
````

````python
numerical_pipeline = Pipeline(
    [("imputer", SimpleImputer(strategy="mean")), ("scaler", StandardScaler())]
)
````

````python
categorical_pipeline = Pipeline(
    [("imputer", SimpleImputer(strategy="most_frequent")), ("encoder", OneHotEncoder())]
)
````

````python
preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_columns),
    ("cat", categorical_pipeline, categorical_columns),
])
````

````python
final_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression())
])
````

````python
final_pipeline.fit(X_train, y_train)
````

````python
y_pred = final_pipeline.predict(X_test)
````

````python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
````
