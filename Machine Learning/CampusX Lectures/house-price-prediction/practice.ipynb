```python
path = "/content/drive/MyDrive/data/House-Prediction_Dataset/data.csv"
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(path)
```

```python
df.head()
```

```python
df[df["price"] == 0]["price"].value_counts()

df = df[df["price"] > 0]
```

```python
X = df.drop("price", axis=1)
Y = df["price"]
```

```python
X.head()
```

```python
X["date"] = pd.to_datetime(df["date"])
# separate day, month, year
X["year"] = X["date"].dt.year
X["month"] = X["date"].dt.month
X["day"] = X["date"].dt.day

# drop date column
X.drop("date", axis=1, inplace=True)
```

```python
X.info()
```

```python
X["street"].nunique()

# street as larget category and act as id
X.drop("street", axis=1, inplace=True)
```

```python
X["statezip"].nunique()
X[["state", "zip"]] = X["statezip"].str.split(" ", n=1, expand=True)
```

```python
X["state"].nunique()
# state has unit value so there is no need of it
X.drop(["state", "statezip"], inplace=True, axis=1)
```

```python
X["zip"] = pd.to_numeric(X["zip"], errors="coerce").astype("int64")
```

```python
X["country"].nunique()
# as counrty as unique value so its can be skipes
X.drop("country", inplace=True, axis=1)
```

```python
X["total_sqft"] = X["sqft_basement"] + X["sqft_above"]
X.drop(["sqft_basement", "sqft_above"], inplace=True, axis=1)
```

```python
X.info()
```

```python
X["city"].nunique()
```

```python
X["year"].nunique()
X.drop("year", axis=1, inplace=True)
```

```python
y_log = np.log1p(Y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, random_state=42, test_size=0.2)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(handle_unknown="ignore")
```

```python
X_train_city = ohe.fit_transform(X_train[["city"]])
X_test_city = ohe.transform(X_test[["city"]])
```

```python
X_train_city
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_num = X_train.drop("city", axis=1)
X_test_num = X_test.drop("city", axis=1)

X_train_scale = scaler.fit_transform(X_train_num)
X_test_scale = scaler.transform(X_test_num)
```

```python
X_train_final = np.hstack((X_train_scale, X_train_city.toarray()))
X_test_final = np.hstack((X_test_scale, X_test_city.toarray()))
```

```python
X_train_final
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train_final, y_train)
y_pred = model.predict(X_test_final)
```

```python
from sklearn.metrics import r2_score, mean_squared_error

r2_score(y_test, y_pred)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=200, random_state=42)

rf.fit(X_train_final, y_train)

y_predic = rf.predict(X_test_final)

r2_score(y_test, y_predic)
```
