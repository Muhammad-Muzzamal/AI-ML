## Handle Missing Values

````python
import numpy as np
student_data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Age": [20, np.nan, 22, 21, None],
    "Math_Score": [85, 78, np.nan, 92, 88],
    "Science_Score": [90, np.nan, 85, None, 87],
    "English_Score": [88, 82, 91, 79, np.nan]
}

df = pd.DataFrame(student_data)
print(df)
print(df.isnull())
print(df.isnull().sum())
````

````python
df.dropna(axis=1, inplace=True)
print(df)
````

````python
# fill null values with default values
df.fillna(0, inplace=True)
print(df)
````

````python
# change in specific column
print(df)
df["Age"].fillna(df["Age"].mean(), inplace=True)
````

````python
print(df)
df.interpolate(method="linear", axis=0, inplace=True)
print("==== Interpolated Data ====")
print(df)
````

## Sorting and Aggregation

````python
# Sorting
data = {
    'name': ['John', 'Emma', 'Michael', 'Sophia', 'William', 'Olivia', 'James', 'Ava'],
    'age': [18, 19, 20, 18, 21, 19, 20, 18],
    'score': [85, 92, 78, 95, 70, 88, 82, 91],
    'grade': ['B', 'A', 'C', 'A', 'D', 'B', 'B', 'A']
}

df = pd.DataFrame(data)
print(df)
df.sort_values(by="age", ascending=True, inplace=True)
print(df)
# sort by two columns
df.sort_values(by=["age", "grade"], ascending=True, inplace=True)
print(df)
````

## Merging and Joining

````python
# DataFrame 1: Student information
df1 = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 4, 5],
        "name": ["John", "Emma", "Michael", "Sophia", "William"],
        "age": [18, 19, 20, 18, 21],
    }
)

# DataFrame 2: Student scores
df2 = pd.DataFrame(
    {
        "student_id": [1, 2, 3, 4, 5],
        "score": [85, 92, 78, 95, 70],
        "grade": ["B", "A", "C", "A", "D"],
    }
)
print("==== INNER JOIN ====")
print(pd.merge(df1, df2, on="student_id"))
print("==== OUTER JOIN ====")
print(pd.merge(df1, df2, on="student_id", how="outer"))
print("==== RIGHT JOIN ====")
print(pd.merge(df1, df2, on="student_id", how="right"))
````

### Concatination
```
vertically (row-wise)
horizontally (col-wise)
```

````python
# DataFrame 1: First batch of students
df1 = pd.DataFrame({
    'name': ['John', 'Emma', 'Michael'],
    'age': [18, 19, 20],
    'score': [85, 92, 78]
})

# DataFrame 2: Second batch of students
df2 = pd.DataFrame({
    'name': ['Sophia', 'William', 'Olivia'],
    'age': [18, 21, 19],
    'score': [95, 70, 88]
})

result = pd.concat([df1, df2], ignore_index=True, axis=1)

print(result)
````
