## Dataset 2: Products


```python
import pandas as pd
import numpy as np
```

```python
df_products = pd.DataFrame({
    'product': ['Laptop', 'Phone', 'Tablet', 'Mouse', 'Keyboard'],
    'price': [800, 500, 300, 20, 50],
    'stock': [15, 30, 25, 100, 75],
    'rating': [4.5, 4.2, 4.0, 4.8, 4.3]
})
print(df_products)
```

```python
# Find products with price < 100
print(df_products[df_products["price"] > 100])
```

```python
# Add total_value column (price * stock)
df_products["total_value"] = df_products["price"] * df_products["stock"]
print(df_products)
```

```python
# Find most expensive product
print(df_products[df_products["price"] == df_products["price"].max()])
```

```python
# Sort by rating (highest first)
df_products.sort_values(by="rating", ascending=False, inplace=True)
print(df_products)
```

```python
# Find average price
print(df_products["price"].mean())
```

```python
# Find products with rating > 4.5
print(df_products[df_products["rating"] > 4.5])
```

```python
# Sort by stock (lowest first)
df_sorted = df_products.sort_values(by="stock", ascending=True, inplace=False)
print(df_sorted)
```

```python
# Find product with highest total_value
print(df_products[df_products["total_value"] == df_products["total_value"].max()])
```
