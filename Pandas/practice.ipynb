## Dataset 1: Simple Student Records (Start Here)


```python
import pandas as pd
import numpy as np
```

```python
student_data = {
    'name': ['John', 'Emma', 'Michael', 'Sophia', 'William', 'Olivia', 'James', 'Ava',
             'Robert', 'Isabella', 'David', 'Mia', 'Richard', 'Charlotte', 'Joseph', 
             'Amelia', 'Thomas', 'Harper', 'Charles', 'Evelyn'],
    'age': [18, 19, 20, 18, 21, 19, 20, 18, 22, 19, 20, 21, 18, 19, 20, 21, 22, 18, 19, 20],
    'math_score': [85, 92, 78, 95, 70, 88, 82, 91, 76, 89, 94, 68, 87, 93, 79, 84, 90, 86, 81, 77],
    'science_score': [88, 90, 82, 94, 75, 91, 85, 89, 80, 92, 96, 72, 86, 95, 81, 88, 93, 84, 79, 82],
    'english_score': [82, 88, 85, 91, 78, 86, 84, 92, 79, 87, 90, 75, 83, 89, 80, 85, 88, 81, 84, 79]
}

df = pd.DataFrame(student_data)
```

```python
# print 1st 3 rows
print(df.head(3))
```

```python
print(df[["name", "math_score"]])
```

```python
# Find students with math > 85
print(df[df["math_score"] > 85])
```

```python
# Add total_score column (sum of 3 subjects)
df["total_score"] = df["math_score"] + df["science_score"] + df["english_score"]
print(df)
```

```python
# Add average column
df["average"] = df["total_score"] / 3
print(df)
```

```python
# Sort by total_score (highest first)
df.sort_values(by="total_score", ascending=False, inplace=True)
print(df)
```

```python
# Find oldest student
print(df[df["age"] == df["age"].max()])
```

```python
# Count students by age
print(df["age"].value_counts())
```

```python
print(df["math_score"].max())
```

```python
print(df["english_score"].mean())
```
