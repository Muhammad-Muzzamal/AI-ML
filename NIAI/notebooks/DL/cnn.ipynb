```python
import re
```

```python
def simple_tokenize(text):
    return re.findall(r'[a-z]+', text.lower())
```

```python
sentence = "The movie wasn't perfect, but it was VERY enjoyable"
tokens = simple_tokenize(sentence)
tokens
```

```python
# padding and trancate
def pad_or_truncate(sequence, max_length, pad_id=0):
    sequence = list(sequence[:max_length])
    return sequence + [pad_id] * (max_length - len(sequence))

examples = [[4, 8], [7, 2, 9, 3, 6], [5, 4, 8, 2, 1, 9]]

for seq in examples:
    print(seq, "->", pad_or_truncate(seq, 5))
```
