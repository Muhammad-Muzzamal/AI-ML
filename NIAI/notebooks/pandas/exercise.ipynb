```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115],

    "Age": [20, 21, 20, 22, 21, 23, 20, 22, 21, 24,
            23, 20, 22, 21, 23],

    "Study_Hours": [4, 6, 3, 8, 5, 7, 2, 9, 6, 4,
                    8, 3, 7, 5, 9],

    "Attendance": [85, 90, 78, 95, 88, 92, 75, 96, 89, 82,
                   94, 80, 91, 87, 97],

    "Assignments": [7, 9, 6, 10, 8, 9, 5, 10, 8, 7,
                    9, 6, 10, 8, 10],

    "Midterm": [72, 85, 68, 91, 79, 88, 61, 94, 83, 70,
                89, 65, 92, 77, 95],

    "Final": [78, 88, 72, 94, 82, 91, 65, 96, 86, 75,
              93, 70, 95, 81, 97]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.groupby("Final")["Age"].mean()
```

```python
df.groupby(["Final", "Midterm"])["Attendance"].mean()
```

```python
df.columns
```

```python
df.groupby("Attendance").agg(
    Study_Hours=("Study_Hours", "mean"),
    Midterm=("Midterm", "mean"),
    Final=("Final", "mean")
)
```

```python
df["Final_mean"] = df.groupby("Attendance")["Final"].transform("mean")
```
