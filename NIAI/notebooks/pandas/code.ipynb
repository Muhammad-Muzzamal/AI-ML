```python
def func(L):
    n = len(L) // 2
    new_l = {}
    for i in L:
        new_l[i] = 0

    for i in L:
        new_l[i] += 1

    for key in new_l:
        if new_l[key] > n:
            print(key)
        
    new_l.items()
```
