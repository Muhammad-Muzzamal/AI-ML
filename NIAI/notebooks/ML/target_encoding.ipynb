```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, RobustScaler

# Load dataset
df = pd.read_csv("gme_sales_dataset.csv")

# Fill missing values
df["Critic_Score"] = df["Critic_Score"].fillna(df["Critic_Score"].median())

# Columns
cols = ["Critic_Score", "Marketing_Spend_Thousands"]

# Create scalers
standard_scaler = StandardScaler()
robust_scaler = RobustScaler()

# Scale the data
standard_scaled = standard_scaler.fit_transform(df[cols])
robust_scaled = robust_scaler.fit_transform(df[cols])

# Convert scaled data to DataFrames
standard_df = pd.DataFrame(
    standard_scaled,
    columns=["Critic_Score", "Marketing_Spend_Thousands"]
)

robust_df = pd.DataFrame(
    robust_scaled,
    columns=["Critic_Score", "Marketing_Spend_Thousands"]
)


# =========================================================
# CRITIC SCORE
# =========================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    range(len(df)),
    df["Critic_Score"],
    label="Before Scaling"
)

plt.scatter(
    range(len(df)),
    standard_df["Critic_Score"],
    label="Standard Scaling"
)

plt.scatter(
    range(len(df)),
    robust_df["Critic_Score"],
    label="Robust Scaling"
)

plt.title("Critic Score — Before vs Standard vs Robust Scaling")
plt.xlabel("Game Index")
plt.ylabel("Value")
plt.legend()
plt.grid(True)

plt.show()


# =========================================================
# MARKETING SPEND
# =========================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    range(len(df)),
    df["Marketing_Spend_Thousands"],
    label="Before Scaling"
)

plt.scatter(
    range(len(df)),
    standard_df["Marketing_Spend_Thousands"],
    label="Standard Scaling"
)

plt.scatter(
    range(len(df)),
    robust_df["Marketing_Spend_Thousands"],
    label="Robust Scaling"
)

plt.title("Marketing Spend — Before vs Standard vs Robust Scaling")
plt.xlabel("Game Index")
plt.ylabel("Value")
plt.legend()
plt.grid(True)

plt.show()
```

# Min Max Scalling`

```python
import pandas as pd

data = {
    "Student": ["Ali", "Ahmed", "Sara", "Usman", "Hassan", "Ayesha", "Bilal", "Zain"],
    "Study_Hours": [2, 4, 6, 8, 10, 12, 14, 16],
    "Exam_Score": [35, 42, 55, 61, 70, 78, 88, 95]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Study_Hours"], df["Exam_Score"])
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["Study_Hours_Scaled"] = scaler.fit_transform(df[["Study_Hours"]])
df["Exam_Score_Scaled"] = scaler.fit_transform(df[["Exam_Score"]])
```

```python
df
```

```python
plt.scatter(
    df["Study_Hours_Scaled"],
    df["Exam_Score_Scaled"]
)
plt.scatter(
    df["Study_Hours"],
    df["Exam_Score"]
)

plt.xlabel("Study Hours (Scaled)")
plt.ylabel("Exam Score (Scaled)")
plt.title("After Min-Max Scaling")



plt.show()
```
