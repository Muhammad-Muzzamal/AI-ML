```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv("used_appliances.csv")
```

```python
df.head()
```

```python
df.nunique()
```

```python
le = LabelEncoder()
```

```python
df["brand_label"] = le.fit_transform(df["brand"])
```

```python
df[['brand', 'brand_label']].head()
```

```python
condition_order = [['Poor', 'Fair', 'Good', 'Excellent']]
```

```python
ord_enc = OrdinalEncoder(categories=condition_order)
```

```python
df["condition_ord"] = ord_enc.fit_transform(df[["condition"]])
```

```python
df[["condition", "condition_ord"]].drop_duplicates().sort_values("condition_ord")
```

```python
ohe = OneHotEncoder(drop="first", handle_unknown="ignore", sparse_output=False)
encoded = ohe.fit_transform(df[["appliance_type"]])
```

```python
ohe_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['appliance_type']))
```

```python
pd.concat([df[['appliance_type']], ohe_df], axis=1).head()
```
