```python
L = ["income", "index", "in"]
```

```python
def func(L):
    if not L:
        return ""

    L.sort()
    
    first = L[0]
    last = L[-1]
    
    i = 0
    while i < len(first) and first[i] == last[i]:
        i += 1
    
    return first[:i]
```

```python
func(L)
```

```python
def func1(L1, L2):
    ans = []
    ptr1 = ptr2 = 0

    while ptr1 < len(L1) and ptr2 < len(L2):
        if L1[ptr1] <= L2[ptr2]:
            ans.append(L1[ptr1])
            ptr1 += 1
        else:
            ans.append(L2[ptr2])
            ptr2 += 1
    
    while ptr1 < len(L1):
        ans.append(L1[ptr1])
        ptr1 += 1
    
    while ptr2 < len(L2):
        ans.append(L2[ptr2])
        ptr2 += 1
    
    return ans


print(func1([1, 2, 3], [1, 2, 3]))
```

```python
L1 = [1, 3, 5]
L2 = [2, 4, 6]

func1(L1, L2)
```
