```python
L = [1,2, 3, 4]

```

```python
def func(L):
    new_list = []
    for i in range(len(L)):
        prod = 1
        for idx, val in enumerate(L):
            if(idx == i):
                continue
            else:
                prod *= val
        
        new_list.append(prod)
    return new_list
        
        
func(L)
        
```
