```python
questions = [
    {"question": "Capital of Pakistan?", "options": ["A) Lahore", "B) Karachi", "C) Islamabad"], "answer": "C"},
    {"question": "2 + 3 = ?", "options": ["A) 4", "B) 5", "C) 6"], "answer": "B"},
    {"question": "Python is a ____ language?", "options": ["A) Programming", "B) Cooking", "C) Driving"], "answer": "A"},
    {"question": "Which symbol is used for comments in Python?", "options": ["A) //", "B) #", "C) --"], "answer": "B"},
    {"question": "len() function is used to find?", "options": ["A) Length", "B) Sum", "C) Average"], "answer": "A"},
]
```

```python
def take_quiz():
    score = 0
    for i, q in enumerate(questions, start=1):
        print(f"\nQ{i} : {q["question"]}")
        for option in q["options"]:
            print(option)
        answer = input("Your answer (A/B/C): ")
        answer = answer.strip().upper()

        if answer == q["answer"]:
            print("Correct!")
            score += 1
        else:
            print(f"Wrong! Correct answer: {q["answer"]}")

    return score
```

```python
print("===== SIMPLE QUIZ SYSTEM =====")

name = input("Enter your name: ")

score = take_quiz()
total_score = len(questions)

print("RESULT".center(30)) 
print("-" * 30) 
print(f"{'Name':<12}: {name}") 
print(f"{'Score':<12}: {score}/{total_score}")
```
