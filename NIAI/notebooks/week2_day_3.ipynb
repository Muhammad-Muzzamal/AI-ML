```python
def my_max(L):
    max_value = L[0]
    for i in L:
       if i > max_value:
           max_value = i
    
    return max_value 
def my_min(L):
    min_value = L[0]
    for i in L:
       if i < min_value:
           min_value = i
    
    return min_value 
```

```python
L = []
for i in range(4):
    L.append(int(input("Enter number")))
    
    
print(L)
print(my_max(L))
print(my_min(L))

```

```python
def remove_duplicate(num):
    map = []
    map.append(num[0])
    for i in range(len(num)):
        if num[i] in map:
            continue
        else:
            map.append(num[i])
    
    print(map)
```

```python
def remove_duplicate1(num):
    map = []
    for i in num:
        if i not in map:
            map.append(i)
    
    print(map)
```

```python
num = []
for i in range(4):
    num.append(int(input("Enter number")))

remove_duplicate1(num)
```
