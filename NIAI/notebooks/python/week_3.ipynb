```python
import numpy as np
```

```python
def func(num):
    L = []
    for i in range(1, num+1):
        if(i % 3 == 0):
            L.append("Fizz")
        if(i % 5 == 0):
            L.append("Buzz")
        if(i % 5 == 0 and i % 3 == 0):
            L.append("FizzBuzz")
        else:
            L.append(str(i))
        
    return L
```

```python
func(15)
```
