```python
def calculate_grade(total_marks, obtained_marks):
    percentage = (obtained_marks / total_marks) * 100

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    return grade
```

```python
def login():
    username = input("Enter username: ")
    password = input("Enter password: ")

    for student in students:

        if student["username"] == username and student["password"] == password:

            print("\n" + "=" * 40)
            print("STUDENT PROFILE".center(40))
            print("=" * 40)

            print(f"ID       : {student['id']}")
            print(f"Name     : {student['name']}")
            print(f"Age      : {student['age']}")
            print(f"Username : {student['username']}")

            print("-" * 40)
            print(f"{"SUBJECT":<18}{"MARKS":<11}{"Grade"}")
            print("-" * 40)
            # print(student["marks"].items())
            total_marks = 0
            for subject, marks in student["marks"].items():
                print(f"{subject:<19}{marks:<12}{calculate_grade(100, marks)}")
                total_marks += marks

            print("-" * 40)
            print(f"{"Total Marks":<18} {total_marks:<12}{calculate_grade(500, total_marks)}")
            print("=" * 40)

            return

    print("\nInvalid username or password!")
```
