```python
# matplotlib
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 8, 9]
plt.plot(x, y)
plt.show()
```

```python
# Bakery Data visualization
weekdays = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
sold_items = [45, 52, 48, 60, 75, 92, 68]

plt.plot(weekdays, sold_items)
plt.xlabel("Days of week")
plt.ylabel("Sold items per day")
plt.title("Bakery Sale Week")

plt.show()
```

## Pyplot Functions

```python
#plt.plot(x, y, color="color name", linestyle="line style", linewidth=value, marker="marker symbol", label="label name")

# SAMPLE DATA 1: Monthly sales for a single product
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales_2024 = [1250, 1180, 1450, 1620, 1890, 2100, 2350, 2280, 2450, 2670, 2900, 3100]
plt.plot(months, sales_2024, color="red", linestyle="--", label="Sales data", marker="o")
plt.xlabel("Months")
plt.ylabel("Sale per month")
plt.title("Monthly sale Data Report")
plt.legend()
plt.grid(color="gray", linestyle=":") 
plt.xlim("Jan", "Aug")
plt.show()
```

```python
# plt.bar(x, height, color="color_name", width=value, )
import matplotlib.pyplot as plt
product = ['A', 'B', 'C', 'D']
sales = [1000, 1500, 800, 1200]
plt.bar(product, sales, color="orange", label="Sales 2025")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.legend()
plt.title("Product sale product")
plt.show()
```
