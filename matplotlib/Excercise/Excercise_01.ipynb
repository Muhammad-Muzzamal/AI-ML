```python
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales = [45000, 52000, 48000, 61000, 55000, 70000,
         65000, 72000, 68000, 80000, 95000, 110000]

expenses = [30000, 33000, 31000, 38000, 35000, 42000,
            40000, 44000, 41000, 48000, 55000, 62000]

products = ['Mobile', 'Laptop', 'AC', 'Fridge', 'TV']
product_sales = [35, 25, 20, 12, 8]  # percentage
```

```python
plt.plot(months, sales, label="Sales trends", marker="o", color="green")
plt.title("Ahmed Electronics - 202 Sales")
plt.show()
```

```python
x = np.arange(len(months))
width = 0.4



plt.bar(x - width/2, sales, label="Sales", width=width)
plt.bar(x + width/2, expenses, label="Expenses", width=width)

plt.legend()

plt.show()
```

```python
max_index = product_sales.index(max(product_sales))
explode= [0.05 if i == max_index else 0 for i in range(len(product_sales))]

plt.pie(product_sales, labels=products, autopct='%.1f%%', explode=explode)



plt.show()
```

```python
plt.fill_between(months, sales, color="blue")
plt.grid()
plt.show()
```

```python
# profit chart

profit = [(sales[i] - expenses[i]) for i in range(len(months))]
color=["green" if x > 20000 else "orange" for x in profit]

plt.bar(months, profit, color=color)
plt.title("Monthly Net Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales = [45000, 52000, 48000, 61000, 55000, 70000,
         65000, 72000, 68000, 80000, 95000, 110000]

expenses = [30000, 33000, 31000, 38000, 35000, 42000,
            40000, 44000, 41000, 48000, 55000, 62000]

products = ['Mobile', 'Laptop', 'AC', 'Fridge', 'TV']
product_sales = [35, 25, 20, 12, 8]

# 🆕 Naya data
customers = [120, 135, 125, 160, 148, 180,
             170, 190, 175, 200, 230, 270]

profit = [s - e for s, e in zip(sales, expenses)]
```

```python
# creating subplots
fig, ax = plt.subplots(2, 2, figsize=(14, 8)) # 2d grid

# top-left
ax[0, 0].plot(months, sales)
# top-right
ax[0, 1].bar(months, expenses)
# bottom-left
ax[1, 0].bar(months, profit)
# Bottom-right: Pie chart (products)
ax[1, 1].pie(product_sales, labels=products, autopct="%.1f%%")

plt.tight_layout()
plt.show()

```

```python
plt.style.use("ggplot")

plt.plot(months, sales, color="green", marker="o", linestyle="--", markersize=8)

for i in range(len(months)):
    plt.text(i, sales[i]+1500, str(sales[i]), ha="center", fontsize=8)

plt.title("Sales Chart")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

ax.bar(months, sales, color="green", label="Sales", alpha=0.7)
ax.set_ylabel("Sales (Rs.)", color="green")
ax.set_title("Sales vs Customer")

ax1 = ax.twinx()
ax1.plot(months, customers, color="red", marker="^", linestyle="--", label="Customers")
ax1.set_ylabel("Customers", color="red")


plt.show()
```
