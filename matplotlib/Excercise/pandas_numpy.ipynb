# Pandas and Numpy

```python
import numpy as np
import pandas as pd

products = ['Mobile', 'Laptop', 'AC', 'Fridge', 'TV', 'Washing Machine']
prices =   [25000,    55000,    40000, 35000,   30000,  45000]
stock =    [50,       30,       20,    15,       25,     10]
sold =     [30,       20,       15,    10,       18,     7]
```

```python
prices_array = np.array(prices)
stock_array = np.array(stock)
sold_array = np.array(sold)

print(prices_array)
print(stock_array)
print(sold_array)
# problem - 01
inventory = prices_array * stock_array
print(inventory)
# Problem - 02
remaining = stock_array - sold_array
print(remaining)
# Problem - 03
df = pd.DataFrame({
    "Products" : products,
    "Prices" : prices_array,
    "Stock" : stock,
    "Sold" : sold
})
print(df[df["Prices"] > 35000])

df["remaining"] = stock_array - sold_array
print(df)
```

```python
import numpy as np
import pandas as pd

products = ['Mobile', 'Laptop', 'AC', 'Fridge', 'TV', 'Washing Machine']
prices =   [25000,    55000,    40000, 35000,   30000,  45000]
stock =    [50,       30,       20,    15,       25,     10]
sold =     [30,       20,       15,    10,       18,     7]
ratings =  [4.5,      4.2,      3.8,   4.0,     4.7,    3.5]

df = pd.DataFrame({
    'Product': products,
    'Price': prices,
    'Stock': stock,
    'Sold': sold,
    'Rating': ratings
})
```

```python
ratings_array = np.array(ratings)
mask = ratings_array > 4.0
print(mask)

# filtered
filtered = ratings_array[mask]
print(filtered)
```

```python
# where class
discount = np.where(prices_array > 35000, prices_array * 0.85, prices_array)
print(discount)
```

```python
df_sorted = df.sort_values(by="Price", ascending=False)
print(df_sorted)
```

```python
nlargest = df.nlargest(3, "Rating")[["Product", "Rating"]]
print(nlargest)
```

```python
df["Status"] = np.select([df["Sold"] >= 20, df["Sold"] >= 10], ["Hot Selling", "Normal"], default="Slow")
print(df[["Product", "Sold", "Status"]])
```
