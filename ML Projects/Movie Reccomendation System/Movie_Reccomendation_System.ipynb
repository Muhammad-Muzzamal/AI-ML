```python
import ast


def convert(obj):
    return ast.literal_eval(obj)
```

```python
df["genres"] = df["genres"].apply(convert)
df["genres"][0]
```

```python
df['cast'] = df["cast"].apply(convert)
type(df["cast"][0])
```

```python
df['keywords'] = df["keywords"].apply(convert)
type(df["keywords"][0])
```

```python
df["crew"] = df["crew"].apply(convert)
type(df["crew"][0])
```

```python
# helper function to fetch name
def fetch_name(obj):
    L = []
    for i in obj:
        L.append(i["name"])
    return L
```

```python
df["genres"] = df["genres"].apply(fetch_name)
```

```python
df["keywords"] = df["keywords"].apply(fetch_name)
df["keywords"]
```

```python
# helper function to fetch name
def fetch_character(obj):
    L = []
    i = 0
    for l in obj:
        if i != 3:
            L.append(l["name"])
            i += 1
        else:
            break
    return L
```

```python
df["cast"]
```

```python
df["cast"] = df["cast"].apply(fetch_character)
```

```python
df["cast"][0]
```

```python
def fetch_producer(obj):
    L = []
    for item in obj:
        if item["job"] == "Producer":
            L.append(item["name"])
    
    return L
```

```python
df["crew"] = df["crew"].apply(fetch_producer)
```

```python
def extract_text(obj):
    L = list(map(lambda x : x, obj.split()))
    return L
```

```python
df["overview"] = df["overview"].apply(extract_text)
```

```python
def remove_space(obj):
    return list(map(lambda x: x.replace(" ", ""), obj))
```

```python
df["cast"] = df["cast"].apply(remove_space)
```

```python
df["crew"] = df["crew"].apply(remove_space)
```

```python
df["genres"] = df["genres"].apply(remove_space)
```

```python
df.head()
```

```python
new_df = df
```

```python
new_df["tags"] = (new_df["genres"]+ new_df["cast"] + new_df["crew"] + new_df["overview"] + df["keywords"])
```

```python
new_df = new_df.drop(columns=["genres", "keywords", "cast", "crew", "overview"])
```

```python
new_df
```

```python
new_df["tags"] = new_df["tags"].apply(lambda x : " ".join(x))
```

```python
new_df["tags"][0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english', max_features=5000)
```

```python
vectors = cv.fit_transform(new_df["tags"]).toarray()
```

```python
cv.get_feature_names_out()
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df["tags"] = new_df["tags"].apply(stem)
```

```python
new_df["tags"][0]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
def recommend(movie):
    movie_index = new_df[new_df["title"] == movie].index[0]
    distance = similarity[movie_index]
    movies_list = sorted(list(enumerate(distance)), reverse=True, key=lambda x:x[1])[1:6]
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Batman Begins")
```
