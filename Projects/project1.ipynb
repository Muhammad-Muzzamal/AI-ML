```python
# Level 1: Beginner (Data Exploration)
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Fatima"],
    "Age": [22, 25, 21, 24],
    "Salary": [50000, 70000, 65000, 80000]
}

df = pd.DataFrame(data)

print(df)

# understand dataset
print(df.head())

# print shape -> return a tuple 
print(df.shape) # (rows, columns)

# print columns
print(df.columns)

print(df.info)

print(df.describe())

```

```python
print(df["Salary"]) # print single columns 
```

```python
print(df[["Salary", "Name"]])
```

```python
# step # 04 filtering
print(df[df["Salary"] > 70000])
```

```python
print(df[(df["Salary"] > 50000) & (df["Age"] > 18)])
```

```python
print(df["Salary"].mean())
```

```python
print(df.sort_values("Salary", ascending=True))
```

```python
# data visualization 
import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Salary"])
plt.title("Employee Series")
plt.xlabel("Name")
plt.ylabel("Salary")
plt.show()
```
